```python
df_millet = df[df['cmname'] == 'Millet - Retail']
```

```python
df_millet.shape
```

```python
df_millet['admname'].value_counts()
```

```python
#regarder comment le prix du mil évolue dans le temps,
prix_moyen_mensuel = df_millet.groupby('date')['price'].mean()
```

```python
import matplotlib.pyplot as plt

prix_moyen_mensuel.plot(figsize=(12, 5), title="Évolution du prix moyen du mil au Sénégal (2000-2020)")
plt.xlabel("Date")
plt.ylabel("Prix (XOF)")
plt.show()
```

```python
#isoler le mois
df_millet['mois'] = df_millet['date'].dt.month
```

```python
#calcule le prix moyen par mois (toutes années confondues)
prix_par_mois = df_millet.groupby('mois')['price'].mean()
prix_par_mois.plot(kind='bar', figsize=(10, 5), title="Prix moyen du mil par mois (toutes années confondues)")
```

```python
prix_par_mois.sort_values()
```

```python
prix_fevrier = prix_par_mois.loc[2]
prix_septembre = prix_par_mois.loc[9]
```

```python
ecart_pct = (prix_septembre - prix_fevrier) / prix_fevrier * 100
```

```python
print(ecart_pct)
```

```python
prix_par_region_mois = df_millet.groupby(['admname', 'mois'])['price'].mean()
```

```python
tableau_regions = prix_par_region_mois.unstack()
```

```python
tableau_regions
```

```python
max_par_region = tableau_regions.max(axis=1)
```

```python
max_par_region
```

```python
min_par_region = tableau_regions.min(axis=1)
min_par_region
```

```python
ecart_pct_regions = (max_par_region - min_par_region) / min_par_region * 100
ecart_pct_regions
```

```python
ecart_pct_regions.sort_values(ascending=False)
```
